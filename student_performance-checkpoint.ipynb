# So sánh điểm trung bình giữa nhóm bị khuyết tật và không bị

- Ta muốn kiểm tra xem liệu những học sinh bị khuyết tật điểm trung bình có thấp hơn những học sinh bình thường hay không.
- Đầu tiên, vẽ biểu đồ phân bố mật độ điểm thi giữa hai nhóm là những người bị khuyết tật và những người không bị.

```python
student_disa = student.select('Learning_Disabilities', 'Exam_Score')
student_disa.hist('Exam_Score', bins = np.arange(50, 100, 1), group = 'Learning_Disabilities')
```

Nhìn vào biểu đồ trên thấy sự khác biệt nhẹ giữa phân phối của những người bị khuyết tật và những người không bị. Có thể thấy rằng những người không bị khuyết tật dường như có điểm số trung bình cao hơn nhóm còn lại.

```python
mean_score_tbl = student_disa.group('Learning_Disabilities', np.average)
mean_score_tbl
```

```python
mean_score = mean_score_tbl.column(1)
obs_diff = mean_score.item(0) - mean_score.item(1)
obs_diff
```

Điểm chênh lệch điểm trung bình giữa hai nhóm khoảng 1.08, vậy sự chênh lệch này là ngẫu nhiên hay có ý nghĩa thống kê? Ta sẽ đi kiểm định giả thuyết
- H0: Điểm trung bình giữa hai nhóm như nhau sự khác biệt quan sát được chi là ngẫu nhiên.
- H1: Điểm trung bình của nhóm người không bị khuyết tật cao hơn nhóm còn lại.

Thống kê kiểm định sẽ là sự khác biệt giữa điểm trung bình của nhóm người không bị khuyết tật và nhóm bị khuyết tật

```python
def one_simulate():
    shuffled_label = student_disa.sample(with_replacement = False).column('Learning_Disabilities')
    shuffled_tbl = student_disa.select('Exam_Score').with_column(
        'Learning_Disabilities', shuffled_label
    )
    mean_score_shuffled_tbl = shuffled_tbl.group('Learning_Disabilities', np.average)
    mean_score_shuffled = mean_score_shuffled_tbl.column('Exam_Score average')
    return mean_score_shuffled.item(0) - mean_score_shuffled.item(1)
```

```python
repititions = 5000
differences = make_array()

for _ in range(repititions):
    new_difference = one_simulate()
    differences = np.append(differences, new_difference)
```

```python
Table().with_column('Difference Between Group Means', differences).hist()
print('Observed Difference:', obs_diff)
```

```python
p_value = np.sum(differences >= obs_diff) / 5000
p_value
```

Vì p_value < 0.05 nên ta có đủ cơ sở để bác bỏ giả thuyết H0 và ủng hộ giả thuyết H1.
Vậy điểm trung bình của nhóm học sinh không bị khuyết tật cao hơn điểm trung bình của nhóm bị khuyết tật có ý nghĩa thống kê.

# Dự đoán điểm số của học sinh

Ta sẽ xây dựng mô hình dự đoán điểm số của học sinh dựa trên các feature, để đơn giản, ta sẽ chỉ chọn các feature ở dạng numerical để phân tích:
- Hours_Studied
- Attendance
- Sleep_Hours
- Previous_Scores
- Tutoring_Sessions
- Physical_Activity

```python
student_rg = student.select('Hours_Studied', 'Attendance', 'Sleep_Hours', 'Previous_Scores', 'Tutoring_Sessions', 'Physical_Activity', 'Exam_Score')
student_rg
```

Trước hết ta định nghĩa hàm để tính hệ số tương quan giữa x và y.

```python
def standard_unit(x):
    return (x - np.mean(x)) / np.std(x)

def corr(x, y):
    return np.mean(standard_unit(x) * standard_unit(y))
```

```python
corr(student_rg.column('Hours_Studied'), student_rg.column('Exam_Score'))
```

Tính corr giữa các thuộc tính và "Exam_Score".

```python
for col in student_rg.drop('Exam_Score').labels:
    print('Correlation of', col, 'and Exam_Score:\t', corr(student_rg.column(col), student_rg.column('Exam_Score')))
```

Chia dữ liệu thành tập train và test với độ dài bằng nhau.

```python
train, test = student_rg.split(3304)
print(train.num_rows, 'training and', test.num_rows, 'test instances.')
```

Chạy thử dự đoán với các slope ngẫu nhiên.

```python
def predict(slopes, row):
    return sum(slopes * np.array(row))

example_row = test.drop('Exam_Score').row(0)
print('Predicting exam score for:', example_row)
example_slopes = np.random.normal(10, 1, len(example_row))
print('Using slopes:', example_slopes)
print('Result:', predict(example_slopes, example_row))
```

```python
print('Actual exam score:', test.column('Exam_Score').item(0))
print('Predicted sale price using random slopes:', predict(example_slopes, example_row))
```

Tiếp theo là định nghĩa hàm rmse.

```python
train_scores = train.column('Exam_Score')
train_attributes = train.drop('Exam_Score')

def rmse(slopes, attributes, scores):
    errors = []
    for i in np.arange(len(scores)):
        predicted = predict(slopes, attributes.row(i))
        actual = scores.item(i)
        errors.append((predicted - actual) ** 2)
    return np.mean(errors) ** 0.5

def rmse_train(slopes):
    return rmse(slopes, train_attributes, train_scores)

print('RMSE of all training examples using random slopes:', rmse_train(example_slopes))
```

```python
best_slopes = minimize(rmse_train, start=example_slopes, smooth=True, array=True)
print('The best slopes for the training set:')
Table(train_attributes.labels).with_row(list(best_slopes)).show()
print('RMSE of all training examples using the best slopes:', rmse_train(best_slopes))
```

Chạy thử với tập test.

```python
test_scores = test.column('Exam_Score')
test_attributes = test.drop('Exam_Score')

def rmse_test(slopes):
    return rmse(slopes, test_attributes, test_scores)

rmse_linear = rmse_test(best_slopes)
print('Test set RMSE for multiple linear regression:', rmse_linear)
```

```python
def fit(row):
    return sum(best_slopes * np.array(row))

test.with_column('Fitted', test.drop('Exam_Score').apply(fit)).scatter('Fitted', 'Exam_Score')
```

Vẽ đồ thị residual để đánh giá mô hình.

```python
test.with_column('Residual', test_scores - test.drop('Exam_Score').apply(fit)).scatter('Exam_Score', 'Residual')
plots.plot([50, 105], [0, 0])
```

Dựa vào đồ thị của residual ta thấy rằng mô hình có vẻ phù hợp khi dự đoán với những học sinh có điểm thấp hơn 80 và bị underestimate đối với những học sinh có điểm từ 80 trờ lên.

Để tìm lý do tại sao mô hình không dự đoán tốt với những học sinh có điểm >= 80 ta sẽ lọc ra các học sinh đó và tính hệ số tương quan giữa các thuộc tính với "Exam_Score".
